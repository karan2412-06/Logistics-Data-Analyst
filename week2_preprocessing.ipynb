# Week 2 – Logistics Data Preprocessing
This notebook demonstrates the main preprocessing steps.

```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

df = pd.read_csv('../data/logistics_data.csv')
df.head()
```

```python
print(df.info())
print(df.isnull().sum())
print('Duplicate rows:', df.duplicated().sum())
```

```python
df = df.drop_duplicates().copy()
cols = ['Distance_km','Delivery_Time_hr','Inventory_Units','Shipping_Cost']
for col in cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
    df[col] = df[col].fillna(df[col].median())
df.isnull().sum()
```

```python
q1 = df['Shipping_Cost'].quantile(.25)
q3 = df['Shipping_Cost'].quantile(.75)
iqr = q3 - q1
df['Cost_Outlier'] = (df['Shipping_Cost'] < q1-1.5*iqr) | (df['Shipping_Cost'] > q3+1.5*iqr)
df[['Shipment_ID','Shipping_Cost','Cost_Outlier']]
```

```python
scaler = MinMaxScaler()
scale_cols = ['Distance_km','Delivery_Time_hr','Shipping_Cost']
df[scale_cols] = scaler.fit_transform(df[scale_cols])
df.head()
```
